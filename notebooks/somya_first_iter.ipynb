```python
import requests
from bs4 import BeautifulSoup

url = "https://datascience.uchicago.edu/education/masters-programs/ms-in-applied-data-science/"

response = requests.get(url)

print(response.status_code)
```

```python
from bs4 import BeautifulSoup

soup = BeautifulSoup(response.text, "html.parser")
```

```python
print(soup.title.text)
```

```python
headings = soup.find_all(["h1", "h2"])

for h in headings:
    print(h.get_text(strip=True))
```

```python
paragraphs = soup.find_all("p")

for p in paragraphs:
    text = p.get_text(strip=True)

    if text:
        print(text)
```

```python
links = soup.find_all("a", href=True)

for link in links:
    print(link.get_text(strip=True), "->", link["href"])
```

```python
with open("uchicago_msads.txt", "w", encoding="utf-8") as f:

    for p in soup.find_all("p"):
        text = p.get_text(strip=True)

        if text:
            f.write(text + "\n\n")

print("Saved!")
```

```python
import requests
from bs4 import BeautifulSoup

url = "https://datascience.uchicago.edu/education/masters-programs/ms-in-applied-data-science/"

response = requests.get(url)
soup = BeautifulSoup(response.text, "html.parser")

with open(r"C:\Users\pujav\Downloads\uchicago_complete.txt", "w", encoding="utf-8") as f:

    for tag in soup.find_all(["h1", "h2", "h3", "p"]):
        text = tag.get_text(" ", strip=True)

        if text:
            f.write(text + "\n\n")

print("Done!")
```

```python
with open(r"C:\Users\pujav\Downloads\uchicago_links.txt", "w", encoding="utf-8") as f:

    for link in soup.find_all("a", href=True):
        text = link.get_text(" ", strip=True)
        href = link["href"]

        f.write(f"{text} --> {href}\n")

print("Links saved!")
```
